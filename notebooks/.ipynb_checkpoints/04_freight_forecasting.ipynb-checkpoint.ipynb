## FreightIQ — Model 1
### Step 1: Load Feature-Engineered Dataset

Load the cleaned feature-engineered dataset from the previous notebook.
This dataset will be used for time-based splitting and model development.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 1: LOAD FEATURE-ENGINEERED DATASET
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Project configuration
# ------------------------------------------------------------

PROJECT_DIR = (
    Path.home()
    / "OneDrive"
    / "Desktop"
    / "Freight rate forecasting ML project"
)

DATA_PATH = (
    PROJECT_DIR
    / "data"
    / "processed"
    / "model1_feature_engineered.csv"
)

print("Project exists:", PROJECT_DIR.exists())
print("Dataset exists:", DATA_PATH.exists())

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found at:\n{DATA_PATH}"
    )

# ------------------------------------------------------------
# Load feature-engineered dataset
# ------------------------------------------------------------

model_data = pd.read_csv(
    DATA_PATH,
    parse_dates=["Date"]
)

# Ensure chronological order
model_data = (
    model_data
    .sort_values("Date")
    .reset_index(drop=True)
)

print("=" * 70)
print("FREIGHTIQ — MODEL 1 FORECASTING DATASET")
print("=" * 70)

print("Shape:", model_data.shape)

print(
    "Date Range:",
    model_data["Date"].min().date(),
    "→",
    model_data["Date"].max().date()
)

print("\nMissing Values:", model_data.isna().sum().sum())

print("\nColumns:", len(model_data.columns))

print("\nFirst 5 Rows:")
display(model_data.head())
```

## Model 1 — Step 2: Time-Based Train/Test Split

Since FreightIQ is a time-series forecasting problem, the dataset must be
split chronologically.

The earlier 80% of observations will be used for training, while the latest
20% will be reserved for testing.

No random shuffling is performed because future observations must never be
used to train the model.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 2: TIME-BASED TRAIN/TEST SPLIT
# ============================================================

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

TRAIN_RATIO = 0.80

# ------------------------------------------------------------
# Calculate split point
# ------------------------------------------------------------

split_index = int(len(model_data) * TRAIN_RATIO)

# ------------------------------------------------------------
# Perform chronological split
# ------------------------------------------------------------

train_data = model_data.iloc[:split_index].copy()
test_data = model_data.iloc[split_index:].copy()

# ------------------------------------------------------------
# Separate features and target
# ------------------------------------------------------------

FEATURE_COLUMNS = [
    column
    for column in model_data.columns
    if column not in ["Date", "BDI_Target"]
]

TARGET_COLUMN = "BDI_Target"

X_train = train_data[FEATURE_COLUMNS]
y_train = train_data[TARGET_COLUMN]

X_test = test_data[FEATURE_COLUMNS]
y_test = test_data[TARGET_COLUMN]

# ------------------------------------------------------------
# Display split information
# ------------------------------------------------------------

print("=" * 70)
print("TIME-BASED TRAIN/TEST SPLIT")
print("=" * 70)

print("Total observations:", len(model_data))

print("\nTraining observations:", len(train_data))
print(
    "Training date range:",
    train_data["Date"].min().date(),
    "→",
    train_data["Date"].max().date()
)

print("\nTesting observations:", len(test_data))
print(
    "Testing date range:",
    test_data["Date"].min().date(),
    "→",
    test_data["Date"].max().date()
)

print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("\ny_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nRandom shuffling: NO")
print("Split method: Chronological 80/20")
```

## Model 1 — Step 3: Naive Baseline

The Naive forecasting model uses the current BDI value as the forecast
for the future BDI target.

For a 30-observation forecasting horizon:

    Forecasted BDI = Current BDI

This baseline provides a simple reference point for evaluating the
performance of more advanced forecasting models.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 3: NAIVE BASELINE MODEL
# ============================================================

# Naive forecast:
# Use the current BDI value to predict the future BDI.
y_pred_naive = test_data["BDI"].values

# Actual future BDI values
y_actual = y_test.values

print("=" * 70)
print("NAIVE BASELINE MODEL")
print("=" * 70)

print("Forecast rule:")
print("Future BDI = Current BDI")

print("\nNumber of test observations:", len(y_actual))

print("\nFirst 10 predictions:")
naive_results = pd.DataFrame({
    "Date": test_data["Date"].values,
    "Current_BDI": test_data["BDI"].values,
    "Actual_BDI": y_actual,
    "Naive_Prediction": y_pred_naive
})

display(naive_results.head(10))
```

## Model 1 — Step 4: Evaluate Naive Baseline

The Naive Baseline provides a reference performance level for the forecasting
problem.

We will evaluate its predictions using three standard regression metrics:

- **MAE (Mean Absolute Error):** Average absolute difference between predicted
  and actual BDI values.
- **RMSE (Root Mean Squared Error):** Measures prediction error while giving
  greater importance to larger errors.
- **MAPE (Mean Absolute Percentage Error):** Measures the average percentage
  difference between predicted and actual BDI values.

These results will later be compared with the other forecasting models.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 4: EVALUATE NAIVE BASELINE
# ============================================================

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)
import numpy as np

# ------------------------------------------------------------
# Calculate evaluation metrics
# ------------------------------------------------------------

mae_naive = mean_absolute_error(
    y_actual,
    y_pred_naive
)

rmse_naive = np.sqrt(
    mean_squared_error(
        y_actual,
        y_pred_naive
    )
)

# MAPE calculation
mape_naive = np.mean(
    np.abs(
        (y_actual - y_pred_naive) / y_actual
    )
) * 100

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("NAIVE BASELINE — PERFORMANCE")
print("=" * 70)

print(f"MAE  : {mae_naive:.2f}")
print(f"RMSE : {rmse_naive:.2f}")
print(f"MAPE : {mape_naive:.2f}%")
```

## Model 1 — Step 5: Moving Average Baseline

The Moving Average baseline predicts the future BDI using the average of
the most recent 30 observed BDI values.

For each test observation:

    Prediction = Average of the previous 30 BDI observations

This provides a second simple benchmark for comparison with the Naive model
before applying advanced statistical and machine learning models.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 5: MOVING AVERAGE BASELINE
# ============================================================

# Moving Average window
MOVING_AVERAGE_WINDOW = 30

# ------------------------------------------------------------
# Create Moving Average predictions
# ------------------------------------------------------------

# Use the complete chronological BDI series
bdi_series = model_data["BDI"]

# Calculate the rolling mean using only previous observations.
# shift(1) prevents the current BDI value from being used
# in its own prediction.
moving_average = (
    bdi_series
    .shift(1)
    .rolling(
        window=MOVING_AVERAGE_WINDOW
    )
    .mean()
)

# Extract predictions corresponding to the test period
y_pred_ma = moving_average.iloc[
    split_index:
].values

# ------------------------------------------------------------
# Display predictions
# ------------------------------------------------------------

print("=" * 70)
print("MOVING AVERAGE BASELINE")
print("=" * 70)

print(
    "Moving Average window:",
    MOVING_AVERAGE_WINDOW,
    "observations"
)

print(
    "Number of test observations:",
    len(y_pred_ma)
)

# Create result table
moving_average_results = pd.DataFrame({
    "Date": test_data["Date"].values,
    "Current_BDI": test_data["BDI"].values,
    "Actual_BDI": y_actual,
    "Moving_Average_Prediction": y_pred_ma
})

print("\nFirst 10 predictions:")
display(moving_average_results.head(10))
```

## Model 1 — Step 6: Evaluate Moving Average Baseline

The Moving Average baseline will be evaluated using the same metrics used
for the Naive baseline:

- MAE (Mean Absolute Error)
- RMSE (Root Mean Squared Error)
- MAPE (Mean Absolute Percentage Error)

These results will allow us to compare the Moving Average baseline with
the Naive baseline.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 6: EVALUATE MOVING AVERAGE BASELINE
# ============================================================

# ------------------------------------------------------------
# Calculate evaluation metrics
# ------------------------------------------------------------

mae_ma = mean_absolute_error(
    y_actual,
    y_pred_ma
)

rmse_ma = np.sqrt(
    mean_squared_error(
        y_actual,
        y_pred_ma
    )
)

mape_ma = np.mean(
    np.abs(
        (y_actual - y_pred_ma) / y_actual
    )
) * 100

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("MOVING AVERAGE BASELINE — PERFORMANCE")
print("=" * 70)

print(f"MAE  : {mae_ma:.2f}")
print(f"RMSE : {rmse_ma:.2f}")
print(f"MAPE : {mape_ma:.2f}%")
```

## Model 1 — Step 7: ARIMA/SARIMA Preparation

ARIMA (AutoRegressive Integrated Moving Average) is a time-series
forecasting model that uses historical values and patterns in the target
variable to forecast future values.

For FreightIQ, ARIMA/SARIMA will be evaluated as an advanced forecasting
model against the Naive and Moving Average baselines.

Before training the model, the BDI series is prepared in chronological
order and checked for stationarity.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 7: ARIMA/SARIMA PREPARATION
# ============================================================

# ------------------------------------------------------------
# Prepare BDI time series
# ------------------------------------------------------------

# Use only the training period to prevent future information
# from entering the ARIMA model.
bdi_train = train_data[
    ["Date", "BDI"]
].copy()

# Set Date as the time-series index
bdi_train = (
    bdi_train
    .set_index("Date")
    .sort_index()
)

# Extract the BDI series
bdi_train_series = bdi_train["BDI"]

print("=" * 70)
print("ARIMA/SARIMA — BDI TRAINING SERIES")
print("=" * 70)

print("Number of observations:", len(bdi_train_series))

print(
    "Date range:",
    bdi_train_series.index.min().date(),
    "→",
    bdi_train_series.index.max().date()
)

print("\nMissing values:", bdi_train_series.isna().sum())

print("\nFirst 5 observations:")
display(bdi_train_series.head())

print("\nLast 5 observations:")
display(bdi_train_series.tail())
```

## Model 1 — Step 8: ARIMA Stationarity Check

ARIMA models require the time series to be stationary or transformed into
a stationary series.

We will use the Augmented Dickey-Fuller (ADF) test to check whether the
training BDI series is stationary.

Interpretation:

- **p-value < 0.05:** The series is considered stationary.
- **p-value >= 0.05:** The series is considered non-stationary and
  differencing may be required.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 8: ARIMA STATIONARITY CHECK
# ============================================================

from statsmodels.tsa.stattools import adfuller

# ------------------------------------------------------------
# Perform Augmented Dickey-Fuller (ADF) test
# ------------------------------------------------------------

adf_result = adfuller(
    bdi_train_series.dropna()
)

adf_statistic = adf_result[0]
p_value = adf_result[1]
used_lags = adf_result[2]
observations = adf_result[3]

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("AUGMENTED DICKEY-FULLER (ADF) TEST")
print("=" * 70)

print(f"ADF Statistic       : {adf_statistic:.4f}")
print(f"p-value             : {p_value:.6f}")
print(f"Number of lags      : {used_lags}")
print(f"Observations used   : {observations}")

print("\nConclusion:")

if p_value < 0.05:
    print("The BDI series is stationary (p-value < 0.05).")
else:
    print("The BDI series is non-stationary (p-value >= 0.05).")
```

## Model 1 — Step 9: First-Order Differencing

The Augmented Dickey-Fuller test showed that the original BDI series is
non-stationary.

First-order differencing is applied to remove the changing trend from the
series.

The differenced value represents the change in BDI from one observation
to the next:

    Differenced BDI = Current BDI − Previous BDI

After differencing, the ADF test will be performed again to check whether
the transformed series is stationary.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 9: FIRST-ORDER DIFFERENCING
# ============================================================

# ------------------------------------------------------------
# Apply first-order differencing
# ------------------------------------------------------------

bdi_train_diff = bdi_train_series.diff().dropna()

# ------------------------------------------------------------
# Display differenced series information
# ------------------------------------------------------------

print("=" * 70)
print("FIRST-ORDER DIFFERENCING")
print("=" * 70)

print("Original observations:", len(bdi_train_series))
print("Differenced observations:", len(bdi_train_diff))

print("\nFirst 10 differenced values:")
display(bdi_train_diff.head(10))

# ------------------------------------------------------------
# Perform ADF test on differenced series
# ------------------------------------------------------------

adf_diff_result = adfuller(
    bdi_train_diff
)

adf_diff_statistic = adf_diff_result[0]
adf_diff_pvalue = adf_diff_result[1]
adf_diff_lags = adf_diff_result[2]
adf_diff_observations = adf_diff_result[3]

print("\n" + "=" * 70)
print("ADF TEST AFTER FIRST-ORDER DIFFERENCING")
print("=" * 70)

print(f"ADF Statistic       : {adf_diff_statistic:.4f}")
print(f"p-value             : {adf_diff_pvalue:.6f}")
print(f"Number of lags      : {adf_diff_lags}")
print(f"Observations used   : {adf_diff_observations}")

print("\nConclusion:")

if adf_diff_pvalue < 0.05:
    print("The differenced BDI series is stationary.")
    print("ARIMA differencing order: d = 1")
else:
    print("The differenced BDI series is still non-stationary.")
    print("Further differencing may be required.")
```

## Model 1 — Step 10: Identify ARIMA p and q

After first-order differencing, the BDI series is stationary with d = 1.

We now analyze the Autocorrelation Function (ACF) and Partial Autocorrelation
Function (PACF) to identify suitable values for the ARIMA p and q parameters.

- ACF helps identify the Moving Average (q) component.
- PACF helps identify the AutoRegressive (p) component.

The plots will be used to select reasonable candidate values for the ARIMA model.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 10: ACF AND PACF ANALYSIS
# ============================================================

from statsmodels.graphics.tsaplots import (
    plot_acf,
    plot_pacf
)
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

# Number of lags to examine
ACF_PACF_LAGS = 30

# ------------------------------------------------------------
# ACF Plot
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plot_acf(
    bdi_train_diff,
    lags=ACF_PACF_LAGS,
    ax=plt.gca()
)

plt.title("BDI Differenced Series — ACF")
plt.xlabel("Lag")
plt.ylabel("Autocorrelation")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# PACF Plot
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plot_pacf(
    bdi_train_diff,
    lags=ACF_PACF_LAGS,
    ax=plt.gca(),
    method="ywm"
)

plt.title("BDI Differenced Series — PACF")
plt.xlabel("Lag")
plt.ylabel("Partial Autocorrelation")
plt.tight_layout()
plt.show()
```

“We used ACF and PACF plots after making the BDI series stationary. The ACF showed a gradual decay, while the PACF showed stronger significant lags around the first few lags. This suggested low-order ARIMA models. We therefore considered several candidate p and q combinations and selected the best model based on statistical and forecasting performance.”

Important: We should not finalize p and q from these plots alone. Testing candidate models is the more reliable approach for your project.

## Model 1 — Step 11: Candidate ARIMA Model Selection

The ACF and PACF analysis suggested that low-order ARIMA models are
appropriate candidates.

Since the exact p and q values cannot be determined reliably from the
plots alone, several candidate ARIMA configurations will be tested.

The models will use:

- Differencing order: d = 1
- Different p values
- Different q values

AIC and BIC will be used to compare the candidate models.
Lower AIC/BIC generally indicates a better statistical fit.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 11: CANDIDATE ARIMA MODEL SELECTION
# ============================================================

from statsmodels.tsa.arima.model import ARIMA
import warnings

# Suppress convergence warnings during candidate testing
warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# Candidate ARIMA configurations
# ------------------------------------------------------------

candidate_orders = [
    (1, 1, 0),
    (2, 1, 0),
    (3, 1, 0),
    (1, 1, 1),
    (2, 1, 1),
    (3, 1, 1)
]

arima_results = []

# ------------------------------------------------------------
# Train and evaluate each candidate
# ------------------------------------------------------------

for order in candidate_orders:

    try:
        model = ARIMA(
            bdi_train_series,
            order=order
        )

        fitted_model = model.fit()

        arima_results.append({
            "ARIMA_Order": str(order),
            "AIC": fitted_model.aic,
            "BIC": fitted_model.bic
        })

    except Exception as error:
        print(
            f"ARIMA{order} failed: {error}"
        )

# ------------------------------------------------------------
# Create comparison table
# ------------------------------------------------------------

arima_comparison = pd.DataFrame(
    arima_results
)

arima_comparison = (
    arima_comparison
    .sort_values("AIC")
    .reset_index(drop=True)
)

print("=" * 70)
print("CANDIDATE ARIMA MODEL COMPARISON")
print("=" * 70)

display(arima_comparison)

print("\nBest candidate according to AIC:")
print(
    arima_comparison.iloc[0]["ARIMA_Order"]
)
```

## Model 1 — Step 12: ARIMA(3,1,1) Forecast

Based on the candidate model comparison, ARIMA(3,1,1) has the lowest AIC
and is selected as the current ARIMA candidate.

The model will now be fitted using only the training BDI data and used to
forecast the test period.

The test data will remain unseen during model fitting so that the forecasting
performance can be evaluated fairly.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 12: ARIMA(3,1,1) FORECAST
# FIXED VERSION
# ============================================================

# ------------------------------------------------------------
# Define selected ARIMA order
# ------------------------------------------------------------

ARIMA_ORDER = (3, 1, 1)

# ------------------------------------------------------------
# Fit ARIMA using BDI values only
# This avoids the irregular-date-index issue.
# ------------------------------------------------------------

arima_model = ARIMA(
    bdi_train_series.values,
    order=ARIMA_ORDER
)

arima_fitted = arima_model.fit()

# ------------------------------------------------------------
# Forecast the complete test period
# ------------------------------------------------------------

y_pred_arima = arima_fitted.forecast(
    steps=len(test_data)
)

# Convert forecast to NumPy array
y_pred_arima = np.asarray(y_pred_arima)

# ------------------------------------------------------------
# Display forecast information
# ------------------------------------------------------------

print("=" * 70)
print("ARIMA(3,1,1) FORECAST")
print("=" * 70)

print("ARIMA order:", ARIMA_ORDER)
print("Training observations:", len(bdi_train_series))
print("Forecast observations:", len(y_pred_arima))

# ------------------------------------------------------------
# Create forecast comparison table
# ------------------------------------------------------------

arima_results = pd.DataFrame({
    "Date": test_data["Date"].values,
    "Actual_BDI": y_actual,
    "ARIMA_Prediction": y_pred_arima
})

print("\nFirst 10 ARIMA predictions:")
display(arima_results.head(10))
```

## Model 1 — Step 13: Evaluate ARIMA(3,1,1)

The ARIMA(3,1,1) forecasts will be evaluated using the same metrics used
for the baseline models:

- MAE (Mean Absolute Error)
- RMSE (Root Mean Squared Error)
- MAPE (Mean Absolute Percentage Error)

The results will be compared with the Naive and Moving Average baselines.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 13: EVALUATE ARIMA(3,1,1)
# ============================================================

# ------------------------------------------------------------
# Calculate evaluation metrics
# ------------------------------------------------------------

mae_arima = mean_absolute_error(
    y_actual,
    y_pred_arima
)

rmse_arima = np.sqrt(
    mean_squared_error(
        y_actual,
        y_pred_arima
    )
)

mape_arima = np.mean(
    np.abs(
        (y_actual - y_pred_arima) / y_actual
    )
) * 100

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("ARIMA(3,1,1) — PERFORMANCE")
print("=" * 70)

print(f"MAE  : {mae_arima:.2f}")
print(f"RMSE : {rmse_arima:.2f}")
print(f"MAPE : {mape_arima:.2f}%")
```

## Model 1 — Step 14: Ridge Regression

Ridge Regression is a linear machine learning model with L2 regularization.

It is useful for FreightIQ because several input variables, especially the
different freight indices, are strongly correlated.

Ridge Regression helps reduce the effect of multicollinearity while
learning relationships between the engineered features and the future BDI.

The model will be trained only on the training data and evaluated on the
unseen test data.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 14: RIDGE REGRESSION
# ============================================================

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

# ------------------------------------------------------------
# Create Ridge Regression pipeline
# ------------------------------------------------------------

ridge_model = Pipeline([
    # Standardize features because Ridge is scale-sensitive
    ("scaler", StandardScaler()),

    # Ridge regression with L2 regularization
    ("ridge", Ridge(alpha=1.0))
])

# ------------------------------------------------------------
# Train Ridge model
# ------------------------------------------------------------

ridge_model.fit(
    X_train,
    y_train
)

# ------------------------------------------------------------
# Generate predictions on unseen test data
# ------------------------------------------------------------

y_pred_ridge = ridge_model.predict(
    X_test
)

# ------------------------------------------------------------
# Display first predictions
# ------------------------------------------------------------

print("=" * 70)
print("RIDGE REGRESSION")
print("=" * 70)

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))

ridge_results = pd.DataFrame({
    "Date": test_data["Date"].values,
    "Actual_BDI": y_actual,
    "Ridge_Prediction": y_pred_ridge
})

print("\nFirst 10 predictions:")
display(ridge_results.head(10))
```

## Model 1 — Step 15: Evaluate Ridge Regression

The Ridge Regression model will be evaluated using MAE, RMSE, and MAPE.

These metrics will be compared with the Naive, Moving Average, and ARIMA
models to determine whether Ridge Regression improves the forecasting
performance.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 15: EVALUATE RIDGE REGRESSION
# ============================================================

# ------------------------------------------------------------
# Calculate evaluation metrics
# ------------------------------------------------------------

mae_ridge = mean_absolute_error(
    y_actual,
    y_pred_ridge
)

rmse_ridge = np.sqrt(
    mean_squared_error(
        y_actual,
        y_pred_ridge
    )
)

mape_ridge = np.mean(
    np.abs(
        (y_actual - y_pred_ridge) / y_actual
    )
) * 100

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("RIDGE REGRESSION — PERFORMANCE")
print("=" * 70)

print(f"MAE  : {mae_ridge:.2f}")
print(f"RMSE : {rmse_ridge:.2f}")
print(f"MAPE : {mape_ridge:.2f}%")
```

## Model 1 — Step 16: Random Forest Regression

Random Forest is a non-linear machine learning model that combines
multiple decision trees to learn complex relationships between the
engineered features and future BDI.

Unlike Ridge Regression, Random Forest can capture non-linear patterns
and interactions between freight indices, economic variables, lag
features, and rolling statistics.

The model will be trained on the training dataset and evaluated on the
unseen test dataset.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 16: RANDOM FOREST REGRESSION
# ============================================================

from sklearn.ensemble import RandomForestRegressor

# ------------------------------------------------------------
# Create Random Forest model
# ------------------------------------------------------------

rf_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1
)

# ------------------------------------------------------------
# Train Random Forest
# ------------------------------------------------------------

rf_model.fit(
    X_train,
    y_train
)

# ------------------------------------------------------------
# Generate predictions on unseen test data
# ------------------------------------------------------------

y_pred_rf = rf_model.predict(
    X_test
)

# ------------------------------------------------------------
# Display first predictions
# ------------------------------------------------------------

print("=" * 70)
print("RANDOM FOREST REGRESSION")
print("=" * 70)

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))
print("Number of trees:", 300)

rf_results = pd.DataFrame({
    "Date": test_data["Date"].values,
    "Actual_BDI": y_actual,
    "Random_Forest_Prediction": y_pred_rf
})

print("\nFirst 10 predictions:")
display(rf_results.head(10))
```

## Model 1 — Step 17: Evaluate Random Forest Regression

The Random Forest model will be evaluated using MAE, RMSE, and MAPE.

Its performance will be compared with the previously evaluated forecasting
models to determine whether the non-linear Random Forest model improves
freight forecasting accuracy.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 17: EVALUATE RANDOM FOREST
# ============================================================

mae_rf = mean_absolute_error(y_actual, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_actual, y_pred_rf))
mape_rf = np.mean(
    np.abs((y_actual - y_pred_rf) / y_actual)
) * 100

print("=" * 70)
print("RANDOM FOREST — PERFORMANCE")
print("=" * 70)

print(f"MAE  : {mae_rf:.2f}")
print(f"RMSE : {rmse_rf:.2f}")
print(f"MAPE : {mape_rf:.2f}%")
```

## Model 1 — Step 18: XGBoost Regression

XGBoost is a gradient-boosting model that can capture complex
non-linear relationships and interactions between the engineered
freight-market features.

The model will be trained on the time-based training dataset and
evaluated on the unseen test dataset.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 18: XGBOOST REGRESSION
# ============================================================

from xgboost import XGBRegressor

# ------------------------------------------------------------
# Create XGBoost model
# ------------------------------------------------------------

xgb_model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

# ------------------------------------------------------------
# Train model
# ------------------------------------------------------------

xgb_model.fit(
    X_train,
    y_train
)

# ------------------------------------------------------------
# Generate predictions
# ------------------------------------------------------------

y_pred_xgb = xgb_model.predict(
    X_test
)

# ------------------------------------------------------------
# Display first predictions
# ------------------------------------------------------------

print("=" * 70)
print("XGBOOST REGRESSION")
print("=" * 70)

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))
print("Number of boosting rounds:", 300)

xgb_results = pd.DataFrame({
    "Date": test_data["Date"].values,
    "Actual_BDI": y_actual,
    "XGBoost_Prediction": y_pred_xgb
})

print("\nFirst 10 predictions:")
display(xgb_results.head(10))
```

## Model 1 — Step 21: Time-Series Cross-Validation

TimeSeriesSplit will be used to validate the forecasting models using
multiple chronological train-validation splits.

Unlike random cross-validation, future observations are never used to
train a model before they occur.

This provides a more reliable estimate of forecasting performance
across different time periods.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 21: TIME-SERIES CROSS-VALIDATION
# ============================================================

from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error

# ------------------------------------------------------------
# Create time-series splits
# ------------------------------------------------------------

tscv = TimeSeriesSplit(n_splits=5)

ridge_cv_scores = []
rf_cv_scores = []
xgb_cv_scores = []

# ------------------------------------------------------------
# Validate the three strongest ML models
# ------------------------------------------------------------

for fold, (train_idx, val_idx) in enumerate(
    tscv.split(X_train), start=1
):

    X_tr = X_train.iloc[train_idx]
    X_val = X_train.iloc[val_idx]

    y_tr = y_train.iloc[train_idx]
    y_val = y_train.iloc[val_idx]

    # --------------------------------------------------------
    # Ridge
    # --------------------------------------------------------

    ridge_cv = Pipeline([
        ("scaler", StandardScaler()),
        ("ridge", Ridge(alpha=1.0))
    ])

    ridge_cv.fit(X_tr, y_tr)
    pred_ridge = ridge_cv.predict(X_val)

    ridge_cv_scores.append(
        np.sqrt(mean_squared_error(y_val, pred_ridge))
    )

    # --------------------------------------------------------
    # Random Forest
    # --------------------------------------------------------

    rf_cv = RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    )

    rf_cv.fit(X_tr, y_tr)
    pred_rf = rf_cv.predict(X_val)

    rf_cv_scores.append(
        np.sqrt(mean_squared_error(y_val, pred_rf))
    )

    # --------------------------------------------------------
    # XGBoost
    # --------------------------------------------------------

    xgb_cv = XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    )

    xgb_cv.fit(X_tr, y_tr)
    pred_xgb = xgb_cv.predict(X_val)

    xgb_cv_scores.append(
        np.sqrt(mean_squared_error(y_val, pred_xgb))
    )

    print(
        f"Fold {fold} completed"
    )

# ------------------------------------------------------------
# Display cross-validation results
# ------------------------------------------------------------

cv_results = pd.DataFrame({
    "Model": [
        "Ridge",
        "Random Forest",
        "XGBoost"
    ],
    "Mean_RMSE": [
        np.mean(ridge_cv_scores),
        np.mean(rf_cv_scores),
        np.mean(xgb_cv_scores)
    ]
})

print("\n" + "=" * 70)
print("TIME-SERIES CROSS-VALIDATION RESULTS")
print("=" * 70)

display(cv_results.sort_values("Mean_RMSE").reset_index(drop=True))
```

## Model 1 — Step 22: Complete Time-Series Validation

The three strongest machine learning models will be compared across
all five time-series folds using MAE, RMSE, and MAPE.

The mean values across the folds will provide a more reliable basis
for final model selection.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 22: COMPLETE TIME-SERIES VALIDATION
# ============================================================

# ------------------------------------------------------------
# Store validation metrics for each model
# ------------------------------------------------------------

cv_metrics = {
    "Ridge": {"MAE": [], "RMSE": [], "MAPE": []},
    "Random Forest": {"MAE": [], "RMSE": [], "MAPE": []},
    "XGBoost": {"MAE": [], "RMSE": [], "MAPE": []}
}

# ------------------------------------------------------------
# Run 5 chronological validation folds
# ------------------------------------------------------------

for fold, (train_idx, val_idx) in enumerate(
    tscv.split(X_train), start=1
):

    X_tr = X_train.iloc[train_idx]
    X_val = X_train.iloc[val_idx]

    y_tr = y_train.iloc[train_idx]
    y_val = y_train.iloc[val_idx]

    # --------------------------------------------------------
    # Ridge
    # --------------------------------------------------------

    ridge_cv = Pipeline([
        ("scaler", StandardScaler()),
        ("ridge", Ridge(alpha=1.0))
    ])

    ridge_cv.fit(X_tr, y_tr)
    pred_ridge = ridge_cv.predict(X_val)

    # --------------------------------------------------------
    # Random Forest
    # --------------------------------------------------------

    rf_cv = RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    )

    rf_cv.fit(X_tr, y_tr)
    pred_rf = rf_cv.predict(X_val)

    # --------------------------------------------------------
    # XGBoost
    # --------------------------------------------------------

    xgb_cv = XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    )

    xgb_cv.fit(X_tr, y_tr)
    pred_xgb = xgb_cv.predict(X_val)

    # --------------------------------------------------------
    # Calculate metrics
    # --------------------------------------------------------

    predictions = {
        "Ridge": pred_ridge,
        "Random Forest": pred_rf,
        "XGBoost": pred_xgb
    }

    for model_name, prediction in predictions.items():

        mae = mean_absolute_error(y_val, prediction)

        rmse = np.sqrt(
            mean_squared_error(y_val, prediction)
        )

        mape = np.mean(
            np.abs(
                (y_val.values - prediction)
                / np.maximum(np.abs(y_val.values), 1e-8)
            )
        ) * 100

        cv_metrics[model_name]["MAE"].append(mae)
        cv_metrics[model_name]["RMSE"].append(rmse)
        cv_metrics[model_name]["MAPE"].append(mape)

# ------------------------------------------------------------
# Calculate mean validation performance
# ------------------------------------------------------------

complete_cv_results = pd.DataFrame({
    "Model": list(cv_metrics.keys()),
    "Mean_MAE": [
        np.mean(cv_metrics[m]["MAE"])
        for m in cv_metrics
    ],
    "Mean_RMSE": [
        np.mean(cv_metrics[m]["RMSE"])
        for m in cv_metrics
    ],
    "Mean_MAPE": [
        np.mean(cv_metrics[m]["MAPE"])
        for m in cv_metrics
    ]
})

complete_cv_results = complete_cv_results.sort_values(
    by="Mean_MAE"
).reset_index(drop=True)

print("=" * 70)
print("COMPLETE TIME-SERIES VALIDATION RESULTS")
print("=" * 70)

display(complete_cv_results)

print(
    "\nBest Mean MAE  :",
    complete_cv_results.loc[
        complete_cv_results["Mean_MAE"].idxmin(), "Model"
    ]
)

print(
    "Best Mean RMSE :",
    complete_cv_results.loc[
        complete_cv_results["Mean_RMSE"].idxmin(), "Model"
    ]
)

print(
    "Best Mean MAPE :",
    complete_cv_results.loc[
        complete_cv_results["Mean_MAPE"].idxmin(), "Model"
    ]
)
```

## Model 1 — Step 23: Final Model Selection

Based on the holdout test performance and five-fold time-series
cross-validation, XGBoost is selected as the final forecasting model.

XGBoost provides the lowest RMSE on the unseen test dataset and
the lowest mean RMSE and MAPE during time-series cross-validation.

Ridge achieved a slightly lower MAE, but XGBoost provides the
stronger overall forecasting performance across the evaluation metrics.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# FREIGHT FORECASTING
# STEP 23: FINAL MODEL SELECTION
# ============================================================

final_model_name = "XGBoost"

final_model = xgb_model

print("=" * 70)
print("FINAL MODEL SELECTION")
print("=" * 70)

print("Selected Model :", final_model_name)

print("\nHoldout Test Performance:")
print(f"MAE  : {mae_xgb:.2f}")
print(f"RMSE : {rmse_xgb:.2f}")
print(f"MAPE : {mape_xgb:.2f}%")

xgb_cv_row = complete_cv_results[
    complete_cv_results["Model"] == "XGBoost"
].iloc[0]

print("\nTime-Series Cross-Validation:")
print(f"Mean MAE  : {xgb_cv_row['Mean_MAE']:.2f}")
print(f"Mean RMSE : {xgb_cv_row['Mean_RMSE']:.2f}")
print(f"Mean MAPE : {xgb_cv_row['Mean_MAPE']:.2f}%")
```

## Model 1 — Step 26: Create Multi-Horizon Forecast Targets

Three forecasting horizons will be created for the FreightIQ model:

- 7 observations ahead
- 14 observations ahead
- 30 observations ahead

Separate target variables will allow us to train dedicated models
for each forecasting horizon.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# STEP 26: CREATE MULTI-HORIZON TARGETS
# ============================================================

# ------------------------------------------------------------
# Create targets for different forecasting horizons
# ------------------------------------------------------------

multi_horizon_data = model_data.copy()

multi_horizon_data["BDI_Target_7"] = (
    multi_horizon_data["BDI"].shift(-7)
)

multi_horizon_data["BDI_Target_14"] = (
    multi_horizon_data["BDI"].shift(-14)
)

multi_horizon_data["BDI_Target_30"] = (
    multi_horizon_data["BDI"].shift(-30)
)

print("=" * 70)
print("MULTI-HORIZON TARGETS")
print("=" * 70)

print("Dataset shape:", multi_horizon_data.shape)

print("\nTarget missing values:")
print(
    multi_horizon_data[
        ["BDI_Target_7", "BDI_Target_14", "BDI_Target_30"]
    ].isna().sum()
)

print("\nLast 5 target observations:")
display(
    multi_horizon_data[
        [
            "Date",
            "BDI",
            "BDI_Target_7",
            "BDI_Target_14",
            "BDI_Target_30"
        ]
    ].tail()
)
```

## Model 1 — Step 27: Prepare Multi-Horizon Training Data

Separate training datasets will be created for the 7-, 14-, and
30-observation forecasting horizons.

Rows without a known future target will be removed separately for
each horizon.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# STEP 27: PREPARE MULTI-HORIZON TRAINING DATA
# ============================================================

# ------------------------------------------------------------
# Feature columns
# ------------------------------------------------------------

feature_columns = X_full.columns.tolist()

# ------------------------------------------------------------
# Create separate datasets for each horizon
# ------------------------------------------------------------

data_7 = multi_horizon_data.dropna(
    subset=["BDI_Target_7"]
).copy()

data_14 = multi_horizon_data.dropna(
    subset=["BDI_Target_14"]
).copy()

data_30 = multi_horizon_data.dropna(
    subset=["BDI_Target_30"]
).copy()

# ------------------------------------------------------------
# Create X and y
# ------------------------------------------------------------

X_7 = data_7[feature_columns]
y_7 = data_7["BDI_Target_7"]

X_14 = data_14[feature_columns]
y_14 = data_14["BDI_Target_14"]

X_30 = data_30[feature_columns]
y_30 = data_30["BDI_Target_30"]

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("=" * 70)
print("MULTI-HORIZON TRAINING DATA")
print("=" * 70)

print(f"7-observation  → X: {X_7.shape}, y: {y_7.shape}")
print(f"14-observation → X: {X_14.shape}, y: {y_14.shape}")
print(f"30-observation → X: {X_30.shape}, y: {y_30.shape}")

print("\nDate ranges:")

print(
    "7-observation :",
    data_7["Date"].min().date(),
    "→",
    data_7["Date"].max().date()
)

print(
    "14-observation:",
    data_14["Date"].min().date(),
    "→",
    data_14["Date"].max().date()
)

print(
    "30-observation:",
    data_30["Date"].min().date(),
    "→",
    data_30["Date"].max().date()
)
```

## Model 1 — Step 28: Time-Based Train/Test Split

Each forecasting horizon will use a chronological 80/20 train-test split.

The earlier observations will be used for training, while the final
20% of observations will be kept completely unseen for evaluation.

No random shuffling will be used because this is a time-series
forecasting problem.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# STEP 29: TIME-BASED TEST SPLIT
# ============================================================

TRAIN_RATIO = 0.80

# ------------------------------------------------------------
# Function for chronological split
# ------------------------------------------------------------

def time_split(X, y):

    split_index = int(len(X) * TRAIN_RATIO)

    X_train_h = X.iloc[:split_index]
    X_test_h = X.iloc[split_index:]

    y_train_h = y.iloc[:split_index]
    y_test_h = y.iloc[split_index:]

    return X_train_h, X_test_h, y_train_h, y_test_h


# ------------------------------------------------------------
# Split 7-observation dataset
# ------------------------------------------------------------

X_train_7, X_test_7, y_train_7, y_test_7 = time_split(
    X_7, y_7
)

# ------------------------------------------------------------
# Split 14-observation dataset
# ------------------------------------------------------------

X_train_14, X_test_14, y_train_14, y_test_14 = time_split(
    X_14, y_14
)

# ------------------------------------------------------------
# Split 30-observation dataset
# ------------------------------------------------------------

X_train_30, X_test_30, y_train_30, y_test_30 = time_split(
    X_30, y_30
)

print("=" * 70)
print("MULTI-HORIZON TIME-BASED TEST SPLIT")
print("=" * 70)

print(
    f"7-observation  → "
    f"Train: {len(X_train_7)} | Test: {len(X_test_7)}"
)

print(
    f"14-observation → "
    f"Train: {len(X_train_14)} | Test: {len(X_test_14)}"
)

print(
    f"30-observation → "
    f"Train: {len(X_train_30)} | Test: {len(X_test_30)}"
)
```

## Model 1 — Step 29: Train Multi-Horizon XGBoost Models

Three separate XGBoost models will be trained using only the
training portion of each chronological split.

The test portions will remain completely unseen until evaluation.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# STEP 29: TRAIN MULTI-HORIZON XGBOOST MODELS
# ============================================================

def create_xgb_model():
    return XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    )


# ------------------------------------------------------------
# Train 7-observation model
# ------------------------------------------------------------

xgb_7_test_model = create_xgb_model()

xgb_7_test_model.fit(
    X_train_7,
    y_train_7
)


# ------------------------------------------------------------
# Train 14-observation model
# ------------------------------------------------------------

xgb_14_test_model = create_xgb_model()

xgb_14_test_model.fit(
    X_train_14,
    y_train_14
)


# ------------------------------------------------------------
# Train 30-observation model
# ------------------------------------------------------------

xgb_30_test_model = create_xgb_model()

xgb_30_test_model.fit(
    X_train_30,
    y_train_30
)


print("=" * 70)
print("MULTI-HORIZON XGBOOST MODELS")
print("=" * 70)

print("7-observation  model : TRAINED")
print("14-observation model : TRAINED")
print("30-observation model : TRAINED")
```

## Model 1 — Step 30: Evaluate Multi-Horizon XGBoost Models

The trained models are evaluated on their unseen chronological test sets
using MAE, RMSE, and MAPE.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# STEP 30: EVALUATE MULTI-HORIZON XGBOOST MODELS
# ============================================================

# Generate predictions
y_pred_7 = xgb_7_test_model.predict(X_test_7)
y_pred_14 = xgb_14_test_model.predict(X_test_14)
y_pred_30 = xgb_30_test_model.predict(X_test_30)


def calculate_metrics(y_actual, y_pred):
    mae = mean_absolute_error(y_actual, y_pred)
    rmse = np.sqrt(mean_squared_error(y_actual, y_pred))
    mape = np.mean(
        np.abs((y_actual - y_pred) / np.maximum(np.abs(y_actual), 1e-8))
    ) * 100

    return mae, rmse, mape


metrics_7 = calculate_metrics(y_test_7, y_pred_7)
metrics_14 = calculate_metrics(y_test_14, y_pred_14)
metrics_30 = calculate_metrics(y_test_30, y_pred_30)


evaluation_results = pd.DataFrame({
    "Forecast Horizon": [
        "7 observations",
        "14 observations",
        "30 observations"
    ],
    "MAE": [
        metrics_7[0],
        metrics_14[0],
        metrics_30[0]
    ],
    "RMSE": [
        metrics_7[1],
        metrics_14[1],
        metrics_30[1]
    ],
    "MAPE (%)": [
        metrics_7[2],
        metrics_14[2],
        metrics_30[2]
    ]
})

print("=" * 70)
print("MULTI-HORIZON XGBOOST EVALUATION")
print("=" * 70)

display(evaluation_results.round(2))
```

## Model 1 — Step 31: TimeSeriesSplit Validation

TimeSeriesSplit is used to validate each forecasting horizon
using multiple chronological training-validation folds.

The final test sets remain untouched.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# STEP 31: TIMESERIES SPLIT VALIDATION
# ============================================================

from sklearn.model_selection import TimeSeriesSplit


def time_series_validation(X, y, horizon_name):
    tscv = TimeSeriesSplit(n_splits=5)

    mae_scores = []
    rmse_scores = []
    mape_scores = []

    for fold, (train_idx, val_idx) in enumerate(tscv.split(X), start=1):

        X_train_fold = X.iloc[train_idx]
        X_val_fold = X.iloc[val_idx]

        y_train_fold = y.iloc[train_idx]
        y_val_fold = y.iloc[val_idx]

        model = create_xgb_model()
        model.fit(X_train_fold, y_train_fold)

        y_val_pred = model.predict(X_val_fold)

        mae, rmse, mape = calculate_metrics(
            y_val_fold,
            y_val_pred
        )

        mae_scores.append(mae)
        rmse_scores.append(rmse)
        mape_scores.append(mape)

    return {
        "Forecast Horizon": horizon_name,
        "Mean MAE": np.mean(mae_scores),
        "Mean RMSE": np.mean(rmse_scores),
        "Mean MAPE (%)": np.mean(mape_scores)
    }


validation_7 = time_series_validation(
    X_train_7, y_train_7, "7 observations"
)

validation_14 = time_series_validation(
    X_train_14, y_train_14, "14 observations"
)

validation_30 = time_series_validation(
    X_train_30, y_train_30, "30 observations"
)


validation_results = pd.DataFrame([
    validation_7,
    validation_14,
    validation_30
])


print("=" * 70)
print("TIMESERIES SPLIT VALIDATION — XGBOOST")
print("=" * 70)

display(validation_results.round(2))
```

## Model 1 — Step 32: Final Model Selection

The forecasting horizon models are compared using both:

1. Chronological test-set performance
2. TimeSeriesSplit validation performance

The selected models will then be retrained on all available data
for their respective forecasting horizons.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# STEP 32: FINAL MODEL SELECTION
# ============================================================

final_model_selection = evaluation_results.merge(
    validation_results,
    on="Forecast Horizon"
)

print("=" * 70)
print("FINAL MULTI-HORIZON MODEL COMPARISON")
print("=" * 70)

display(final_model_selection.round(2))

print("\nSelection basis:")
print("- Test-set MAE, RMSE and MAPE")
print("- TimeSeriesSplit Mean MAE, RMSE and MAPE")
print("- Overall consistency across both evaluations")
```

## Model 1 — Step 33: Retrain Final Multi-Horizon Models

The selected XGBoost architecture is retrained separately for the
7-, 14-, and 30-observation horizons using all available data
with known target values.

These models will be used for the final freight-market forecasts.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# STEP 33: RETRAIN FINAL MULTI-HORIZON MODELS
# ============================================================

final_xgb_7 = create_xgb_model()
final_xgb_7.fit(X_7, y_7)

final_xgb_14 = create_xgb_model()
final_xgb_14.fit(X_14, y_14)

final_xgb_30 = create_xgb_model()
final_xgb_30.fit(X_30, y_30)


print("=" * 70)
print("FINAL MULTI-HORIZON MODELS")
print("=" * 70)

print(f"7-observation  model : TRAINED on {len(X_7)} rows")
print(f"14-observation model : TRAINED on {len(X_14)} rows")
print(f"30-observation model : TRAINED on {len(X_30)} rows")
```

## Model 1 — Step 34: Generate Final Freight Forecasts

The final trained models are used with the latest available feature
values to generate 7-, 14-, and 30-observation BDI forecasts.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# STEP 34: GENERATE FINAL FORECASTS
# ============================================================

# Latest available feature row
latest_features = model_data.iloc[[-1]][feature_columns]

# Generate forecasts
forecast_7 = final_xgb_7.predict(latest_features)[0]
forecast_14 = final_xgb_14.predict(latest_features)[0]
forecast_30 = final_xgb_30.predict(latest_features)[0]

latest_date = model_data["Date"].iloc[-1]
latest_bdi = model_data["BDI"].iloc[-1]

print("=" * 70)
print("FINAL FREIGHT MARKET FORECAST")
print("=" * 70)

print(f"Latest available date : {latest_date.date()}")
print(f"Latest BDI            : {latest_bdi:.2f}")

print("-" * 70)

print(f"7-observation forecast  : {forecast_7:.2f}")
print(f"14-observation forecast : {forecast_14:.2f}")
print(f"30-observation forecast : {forecast_30:.2f}")
```
