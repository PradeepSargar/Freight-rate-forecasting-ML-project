```python
from pathlib import Path

# Project root
PROJECT_DIR = Path.cwd().parent

# Raw freight index directory
FREIGHT_DIR = PROJECT_DIR / "data" / "raw" / "freight_indices"

print("Freight directory:")
print(FREIGHT_DIR)

print("\nFolder exists:", FREIGHT_DIR.exists())
```

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# TASK 1: COMBINE FREIGHT INDEX DATA
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 1. Load BDI parts
# ------------------------------------------------------------

bdi_1 = pd.read_csv(
    FREIGHT_DIR / "Baltic Dry Index" / "BDI_1985-2000.csv"
)

bdi_2 = pd.read_csv(
    FREIGHT_DIR / "Baltic Dry Index" / "BDI_2001-2016.csv"
)

bdi_3 = pd.read_csv(
    FREIGHT_DIR / "Baltic Dry Index" / "BDI_2017-2026.csv"
)

# Combine BDI
bdi = pd.concat(
    [bdi_1, bdi_2, bdi_3],
    ignore_index=True
)

# Keep required columns only
bdi = bdi[["Date", "Price"]].copy()

# Convert data types
bdi["Date"] = pd.to_datetime(
    bdi["Date"],
    errors="coerce"
)

bdi["Price"] = (
    bdi["Price"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)

# Remove duplicate dates
bdi = (
    bdi
    .drop_duplicates(subset="Date")
    .sort_values("Date")
    .reset_index(drop=True)
)

# Rename target
bdi = bdi.rename(columns={"Price": "BDI"})


# ------------------------------------------------------------
# 2. Load BCI parts
# ------------------------------------------------------------

bci_1 = pd.read_csv(
    FREIGHT_DIR / "Baltic Capesize" / "2012-2017.csv"
)

bci_2 = pd.read_csv(
    FREIGHT_DIR / "Baltic Capesize" / "2016-2025.csv"
)

# Combine BCI
bci = pd.concat(
    [bci_1, bci_2],
    ignore_index=True
)

# Keep required columns
bci = bci[["Date", "Price"]].copy()

# Convert data types
bci["Date"] = pd.to_datetime(
    bci["Date"],
    errors="coerce"
)

bci["Price"] = (
    bci["Price"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)

# Remove overlapping duplicate dates
bci = (
    bci
    .drop_duplicates(subset="Date")
    .sort_values("Date")
    .reset_index(drop=True)
)

# Rename
bci = bci.rename(columns={"Price": "BCI"})


# ------------------------------------------------------------
# 3. Load BPI
# ------------------------------------------------------------

bpi = pd.read_csv(
    FREIGHT_DIR / "Baltic Panamax" / "2012-2025.csv"
)

bpi = bpi[["Date", "Price"]].copy()

bpi["Date"] = pd.to_datetime(
    bpi["Date"],
    errors="coerce"
)

bpi["Price"] = (
    bpi["Price"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)

bpi = bpi.rename(columns={"Price": "BPI"})


# ------------------------------------------------------------
# 4. Load BSI
# ------------------------------------------------------------

bsi = pd.read_csv(
    FREIGHT_DIR / "Baltic Supramax" / "2012-25.csv"
)

bsi = bsi[["Date", "Price"]].copy()

bsi["Date"] = pd.to_datetime(
    bsi["Date"],
    errors="coerce"
)

bsi["Price"] = (
    bsi["Price"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)

bsi = bsi.rename(columns={"Price": "BSI"})


# ------------------------------------------------------------
# 5. Load BHSI
# ------------------------------------------------------------

bhsi = pd.read_csv(
    FREIGHT_DIR / "Baltic Handysize" / "2012-2025.csv"
)

bhsi = bhsi[["Date", "Price"]].copy()

bhsi["Date"] = pd.to_datetime(
    bhsi["Date"],
    errors="coerce"
)

bhsi["Price"] = (
    bhsi["Price"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)

bhsi = bhsi.rename(columns={"Price": "BHSI"})


# ============================================================
# 6. Validation
# ============================================================

print("=" * 80)
print("FREIGHT INDEX DATA — CLEANED")
print("=" * 80)

for name, df in {
    "BDI": bdi,
    "BCI": bci,
    "BPI": bpi,
    "BSI": bsi,
    "BHSI": bhsi
}.items():

    print(f"\n{name}")
    print(f"Rows       : {len(df):,}")
    print(f"Start date : {df['Date'].min().date()}")
    print(f"End date   : {df['Date'].max().date()}")
    print(f"Duplicates : {df['Date'].duplicated().sum():,}")
    print(f"Missing    : {df.isna().sum().sum():,}")

print("\n" + "=" * 80)
print("SAMPLE — BDI")
print("=" * 80)

display(bdi.head())

print("\n" + "=" * 80)
print("SAMPLE — BCI")
print("=" * 80)

display(bci.head())
```

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 1: DESCRIPTIVE STATISTICS
# ============================================================

freight_data = {
    "BDI": bdi,
    "BCI": bci,
    "BPI": bpi,
    "BSI": bsi,
    "BHSI": bhsi
}

for name, df in freight_data.items():

    print("\n" + "=" * 70)
    print(f"{name} — DESCRIPTIVE STATISTICS")
    print("=" * 70)

    display(
        df[name].describe().to_frame().round(2)
    )
```

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 3: DATE GAP ANALYSIS
# ============================================================

for name, df in freight_data.items():

    dates = df["Date"].sort_values()
    gaps = dates.diff().dt.days

    print("\n" + "=" * 60)
    print(f"{name} — DATE GAP ANALYSIS")
    print("=" * 60)

    print("Largest gap:", gaps.max(), "days")
    print("Number of gaps > 7 days:", (gaps > 7).sum())
```

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# DEFINE COMMON MODELING PERIOD
# ============================================================

COMMON_START = max(
    df["Date"].min()
    for df in freight_data.values()
)

COMMON_END = min(
    df["Date"].max()
    for df in freight_data.values()
)

print("Common Modeling Period:")
print(COMMON_START.date(), "→", COMMON_END.date())
```

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 4: FREIGHT INDEX TRENDS
# ============================================================

import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

for name, df in freight_data.items():

    period_df = df[
        (df["Date"] >= COMMON_START) &
        (df["Date"] <= COMMON_END)
    ]

    plt.plot(
        period_df["Date"],
        period_df[name],
        label=name,
        linewidth=1
    )

plt.title("Freight Index Trends — Common Modeling Period")
plt.xlabel("Date")
plt.ylabel("Index Value")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 5: FREIGHT INDEX CORRELATION
# ============================================================

common_freight = pd.DataFrame({
    "BDI": bdi.set_index("Date")["BDI"],
    "BCI": bci.set_index("Date")["BCI"],
    "BPI": bpi.set_index("Date")["BPI"],
    "BSI": bsi.set_index("Date")["BSI"],
    "BHSI": bhsi.set_index("Date")["BHSI"]
})

common_freight = common_freight.loc[
    COMMON_START:COMMON_END
]

correlation_matrix = common_freight.corr()

display(correlation_matrix.round(3))
```

## EDA Step 7 — Monthly Commodity Data Inspection

The next step is to inspect the monthly Coal and Iron Ore data
that will be used as external market variables in Model 1.

We will extract:
- Australian Coal
- Iron Ore cfr spot

The data will then be converted into a clean monthly time-series format.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 7: EXTRACT COAL & IRON ORE
# ============================================================

commodity_file = (
    PROJECT_DIR / "data" / "raw" / "market"
    / "Monthly Prices Coal + Iron Ore.xlsx"
)

commodity_raw = pd.read_excel(
    commodity_file,
    sheet_name="Monthly Prices",
    header=None
)

# Extract Date, Australian Coal, and Iron Ore
commodity = commodity_raw.iloc[6:, [0, 5, 63]].copy()

commodity.columns = [
    "Date",
    "Coal",
    "Iron_Ore"
]

# Convert values
commodity["Date"] = pd.to_datetime(
    commodity["Date"].astype(str),
    format="%YM%m",
    errors="coerce"
)

commodity["Coal"] = pd.to_numeric(
    commodity["Coal"],
    errors="coerce"
)

commodity["Iron_Ore"] = pd.to_numeric(
    commodity["Iron_Ore"],
    errors="coerce"
)

commodity = commodity.dropna(subset=["Date"])
commodity = commodity.sort_values("Date").reset_index(drop=True)

print("=" * 70)
print("COAL & IRON ORE DATA")
print("=" * 70)

print("Rows:", len(commodity))
print(
    "Date:",
    commodity["Date"].min().date(),
    "→",
    commodity["Date"].max().date()
)

print("\nMissing values:")
print(commodity[["Coal", "Iron_Ore"]].isna().sum())

display(commodity.head())
```

## EDA Step 11 — Daily Data Alignment

The selected freight and market variables have different observation
dates because of weekends, holidays, and different reporting
frequencies.

A common daily date index will therefore be created for the Model 1
dataset.

This step prepares the data for consistent time-series alignment
without introducing random train-test splitting or future information.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 11: CREATE COMMON DAILY DATE INDEX
# ============================================================

# Create a continuous daily date range for the common modeling period.
daily_dates = pd.DataFrame({
    "Date": pd.date_range(
        start=COMMON_START,
        end=COMMON_END,
        freq="D"
    )
})

print("=" * 70)
print("COMMON DAILY DATE INDEX")
print("=" * 70)

print("Rows:", len(daily_dates))
print(
    "Date:",
    daily_dates["Date"].min().date(),
    "→",
    daily_dates["Date"].max().date()
)

print("\nFirst 5 dates:")
display(daily_dates.head())

print("\nLast 5 dates:")
display(daily_dates.tail())
```

## EDA Step 12 — Merge Daily Freight and Market Variables

The daily freight indices and market variables have different
observation dates.

They will now be aligned using the common daily date index.

Only the selected Model 1 variables will be included:

- BDI
- BCI
- BPI
- BSI
- Brent Crude Oil
- USD/INR

BHSI is excluded because it is not part of the finalized Model 1
feature set.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 12: MERGE DAILY FREIGHT & MARKET DATA
# ============================================================

# Start with the continuous daily date index.
model_daily = daily_dates.copy()

# ------------------------------------------------------------
# Merge selected freight indices
# ------------------------------------------------------------

for name in ["BDI", "BCI", "BPI", "BSI"]:
    
    index_data = freight_data[name][["Date", name]].copy()
    
    model_daily = model_daily.merge(
        index_data,
        on="Date",
        how="left"
    )


# ------------------------------------------------------------
# Merge Brent Crude Oil
# ------------------------------------------------------------

model_daily = model_daily.merge(
    brent_model[["Date", "Brent"]],
    on="Date",
    how="left"
)


# ------------------------------------------------------------
# Merge USD/INR
# ------------------------------------------------------------

model_daily = model_daily.merge(
    usd_inr_model[["Date", "USD_INR"]],
    on="Date",
    how="left"
)


# ------------------------------------------------------------
# Sort by date
# ------------------------------------------------------------

model_daily = model_daily.sort_values("Date").reset_index(drop=True)


# ------------------------------------------------------------
# Display merged dataset information
# ------------------------------------------------------------

print("=" * 70)
print("DAILY FREIGHT & MARKET DATA")
print("=" * 70)

print("Shape:", model_daily.shape)

print("\nColumns:")
print(model_daily.columns.tolist())

print("\nMissing values:")
print(model_daily.isna().sum())

print("\nFirst 5 rows:")
display(model_daily.head())
```

## EDA Step 13 — Missing Value Analysis After Daily Merge

After aligning all daily variables to the common date index, we
identify the missing observations created by weekends, holidays,
and different market reporting schedules.

This analysis will determine the appropriate time-series handling
method before adding the monthly commodity variables.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 13: MISSING VALUE ANALYSIS AFTER DAILY MERGE
# ============================================================

print("=" * 70)
print("MISSING VALUE ANALYSIS — DAILY DATA")
print("=" * 70)

# ------------------------------------------------------------
# 1. Missing-value count
# ------------------------------------------------------------

missing_count = model_daily.isna().sum()

print("\nMissing values:")
print(missing_count)


# ------------------------------------------------------------
# 2. Missing-value percentage
# ------------------------------------------------------------

missing_percentage = (
    model_daily.isna().mean() * 100
).round(2)

print("\nMissing percentage:")
print(missing_percentage)


# ------------------------------------------------------------
# 3. Summary table
# ------------------------------------------------------------

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

print("\nMissing-value summary:")
display(missing_summary)
```

## EDA Step 14 — Missing-Value Pattern Analysis

The daily merged dataset contains approximately 30–32% missing
values because the source datasets do not report observations on
every calendar day.

Before applying any filling method, we identify the pattern of
missing values and distinguish normal non-trading days from actual
data gaps.

This prevents inappropriate imputation of the time-series data.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 14: MISSING-VALUE PATTERN ANALYSIS
# ============================================================

# Create a copy for analysis
missing_pattern = model_daily.copy()

# ------------------------------------------------------------
# Identify rows where all market/freight variables are missing
# ------------------------------------------------------------

data_columns = [
    "BDI",
    "BCI",
    "BPI",
    "BSI",
    "Brent",
    "USD_INR"
]

missing_pattern["All_Missing"] = (
    missing_pattern[data_columns].isna().all(axis=1)
)

# ------------------------------------------------------------
# Count dates where all variables are missing
# ------------------------------------------------------------

all_missing_count = missing_pattern["All_Missing"].sum()

print("=" * 70)
print("MISSING-VALUE PATTERN ANALYSIS")
print("=" * 70)

print(
    "\nDates where all freight and market variables are missing:",
    all_missing_count
)

print(
    "Percentage of total dates:",
    round(
        all_missing_count / len(missing_pattern) * 100,
        2
    ),
    "%"
)

# ------------------------------------------------------------
# Display examples of dates where all variables are missing
# ------------------------------------------------------------

print("\nSample dates with all variables missing:")

display(
    missing_pattern.loc[
        missing_pattern["All_Missing"],
        ["Date"]
    ].head(20)
)
```

## EDA Step 15 — Identify Unexpected Weekday Gaps

Most missing observations are expected on Saturdays and Sundays.
However, we must also check for missing observations on weekdays.

This step identifies weekdays where all selected freight and market
variables are simultaneously missing.

Such dates may represent market holidays or genuine data gaps and
should be investigated before applying any imputation.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 15: IDENTIFY UNEXPECTED WEEKDAY GAPS
# ============================================================

# Create a copy for analysis
weekday_gaps = model_daily.copy()

# Identify the day of the week
weekday_gaps["Day_Name"] = weekday_gaps["Date"].dt.day_name()

# Check whether all selected variables are missing
weekday_gaps["All_Missing"] = (
    weekday_gaps[
        ["BDI", "BCI", "BPI", "BSI", "Brent", "USD_INR"]
    ].isna().all(axis=1)
)

# Keep only weekdays where all variables are missing
weekday_gaps = weekday_gaps[
    (weekday_gaps["Date"].dt.weekday < 5) &
    (weekday_gaps["All_Missing"])
].copy()

print("=" * 70)
print("UNEXPECTED WEEKDAY GAPS")
print("=" * 70)

print("Weekdays with all variables missing:", len(weekday_gaps))

print("\nSample weekday gaps:")

display(
    weekday_gaps[
        ["Date", "Day_Name"]
    ].head(30)
)
```

## EDA Step 16 — Remove Non-Trading Days

The missing-value analysis shows that most completely missing dates
correspond to weekends and recognized market holidays.

Since freight indices and market variables are reported primarily
on trading/business days, these non-trading dates do not represent
missing observations.

They will therefore be removed from the modeling dataset rather
than artificially imputed.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 16: REMOVE NON-TRADING DAYS
# ============================================================

# Define the variables used for daily freight-market modeling.
model_columns = [
    "BDI",
    "BCI",
    "BPI",
    "BSI",
    "Brent",
    "USD_INR"
]

# Keep dates where at least one selected variable has an observation.
# Dates where every variable is missing are non-trading/holiday dates.
model_daily_clean = model_daily[
    model_daily[model_columns].notna().any(axis=1)
].copy()

# Reset the index after removing non-trading dates.
model_daily_clean = model_daily_clean.reset_index(drop=True)

print("=" * 70)
print("DAILY MODELING DATA AFTER REMOVING NON-TRADING DAYS")
print("=" * 70)

print("Original rows:", len(model_daily))
print("Clean rows:", len(model_daily_clean))
print("Rows removed:", len(model_daily) - len(model_daily_clean))

print("\nDate range:")
print(
    model_daily_clean["Date"].min().date(),
    "→",
    model_daily_clean["Date"].max().date()
)

print("\nRemaining missing values:")
print(model_daily_clean[model_columns].isna().sum())
```

## EDA Step 17 — Analyze Remaining Variable Gaps

After removing weekends and holidays, some individual variables still
contain missing observations.

We will measure the consecutive missing gaps for each variable to
determine whether they are isolated missing days or longer data gaps.

This will help us choose an appropriate time-series imputation method
without introducing unnecessary distortion.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 17: ANALYZE REMAINING VARIABLE GAPS
# ============================================================

print("=" * 70)
print("REMAINING MISSING-DATA GAP ANALYSIS")
print("=" * 70)

# Variables requiring gap analysis
analysis_columns = [
    "BDI",
    "BCI",
    "BPI",
    "BSI",
    "Brent",
    "USD_INR"
]

# ------------------------------------------------------------
# Analyze consecutive missing observations for each variable
# ------------------------------------------------------------

for column in analysis_columns:

    # Identify missing observations
    missing = model_daily_clean[column].isna()

    # Create groups whenever missing/non-missing status changes
    groups = missing.ne(missing.shift()).cumsum()

    # Calculate lengths of consecutive missing groups
    gap_lengths = (
        missing.groupby(groups)
        .sum()
    )

    # Keep only actual missing groups
    gap_lengths = gap_lengths[gap_lengths > 0]

    print(f"\n{column}")
    print("-" * 40)

    print("Missing observations:", int(missing.sum()))

    if len(gap_lengths) > 0:
        print("Number of missing gaps:", len(gap_lengths))
        print("Largest consecutive gap:", int(gap_lengths.max()))
        print("Average gap length:", round(gap_lengths.mean(), 2))
    else:
        print("No missing gaps")
```

## EDA Step 18 — Fill Short Internal Missing Gaps

The remaining missing observations are short internal gaps caused by
different market reporting schedules and holidays.

Since the maximum gap is limited to 9 observations, linear
time-based interpolation can be used for internal gaps.

Interpolation will only use values surrounding the missing period.
No forward or backward extrapolation will be performed.

This preserves the overall time-series structure while avoiding
unnecessary loss of observations.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 18: INTERPOLATE SHORT INTERNAL GAPS
# ============================================================

# Work on a copy so the original cleaned dataset remains unchanged.
model_daily_interpolated = model_daily_clean.copy()

# ------------------------------------------------------------
# Ensure Date is the DataFrame index
# ------------------------------------------------------------
# Time-based interpolation requires a datetime index.

model_daily_interpolated = (
    model_daily_interpolated
    .set_index("Date")
)

# ------------------------------------------------------------
# Interpolate only internal missing values
# ------------------------------------------------------------
# limit_area="inside" prevents interpolation at the beginning
# or end of the dataset where no surrounding observation exists.

for column in model_columns:

    model_daily_interpolated[column] = (
        model_daily_interpolated[column]
        .interpolate(
            method="time",
            limit_area="inside"
        )
    )

# ------------------------------------------------------------
# Restore Date as a normal column
# ------------------------------------------------------------

model_daily_interpolated = (
    model_daily_interpolated
    .reset_index()
)

# ------------------------------------------------------------
# Check remaining missing values
# ------------------------------------------------------------

print("=" * 70)
print("MISSING VALUES AFTER TIME-BASED INTERPOLATION")
print("=" * 70)

print(
    model_daily_interpolated[model_columns]
    .isna()
    .sum()
)

print("\nDataset shape:", model_daily_interpolated.shape)
```

## EDA Step 19 — Inspect Remaining Missing Values

After time-based interpolation, a small number of observations
remain missing.

These values are likely located at the boundaries of the available
source data, where interpolation cannot be performed because a
previous or next observation is unavailable.

We will inspect their exact dates before deciding how to handle them.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 19: INSPECT REMAINING MISSING VALUES
# ============================================================

print("=" * 70)
print("REMAINING MISSING VALUES AFTER INTERPOLATION")
print("=" * 70)

# Identify rows containing any remaining missing value
remaining_missing = model_daily_interpolated[
    model_daily_interpolated[model_columns].isna().any(axis=1)
].copy()

print("Rows with remaining missing values:", len(remaining_missing))

print("\nDetails:")
display(
    remaining_missing[
        ["Date"] + model_columns
    ]
)
```

## EDA Step 20 — Final Daily Modeling Dataset

After interpolation, six boundary observations remain incomplete.

Because these values occur at the beginning and end of the modeling
period, they cannot be reliably estimated using internal interpolation.

To avoid artificial values, incomplete boundary observations will be
removed.

The resulting dataset will contain complete daily observations for
all selected freight and market variables.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 20: FINAL DAILY MODELING DATASET
# ============================================================

# ------------------------------------------------------------
# Remove rows that still contain missing values.
# ------------------------------------------------------------
# At this stage, remaining missing values are only boundary gaps.
# Removing them avoids artificial forward/backward extrapolation.

model_daily_final = (
    model_daily_interpolated
    .dropna(subset=model_columns)
    .copy()
)

# Reset index after removing incomplete observations
model_daily_final = (
    model_daily_final
    .sort_values("Date")
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# Final dataset summary
# ------------------------------------------------------------

print("=" * 70)
print("FINAL DAILY MODELING DATASET")
print("=" * 70)

print("Rows:", len(model_daily_final))
print("Columns:", len(model_daily_final.columns))

print(
    "Date:",
    model_daily_final["Date"].min().date(),
    "→",
    model_daily_final["Date"].max().date()
)

print("\nMissing values:")
print(model_daily_final[model_columns].isna().sum())

print("\nColumns:")
print(model_daily_final.columns.tolist())

print("\nFirst 5 rows:")
display(model_daily_final.head())

print("\nLast 5 rows:")
display(model_daily_final.tail())
```

## EDA Step 21 — Align Monthly Commodity Data with Daily Dataset

Coal and Iron Ore are available at monthly frequency, while the freight
and market variables are available at daily frequency.

The monthly commodity values will be aligned with the daily dataset
using the corresponding month.

This allows Coal and Iron Ore to be used as external variables in
Model 1 without changing their original monthly observations.

```python
# ============================================================
# FREIGHTIQ — MODEL 1
# EDA STEP 21: MERGE MONTHLY COMMODITY DATA
# ============================================================

# Create a copy of the final daily dataset
model_final = model_daily_final.copy()

# ------------------------------------------------------------
# Create a common monthly key
# ------------------------------------------------------------
# Both datasets will use the first day of the month as the
# alignment key.

model_final["Month"] = (
    model_final["Date"]
    .dt.to_period("M")
)

commodity_merge = commodity_model.copy()

commodity_merge["Month"] = (
    commodity_merge["Date"]
    .dt.to_period("M")
)

# Keep only the required commodity variables
commodity_merge = commodity_merge[
    ["Month", "Coal", "Iron_Ore"]
]

# ------------------------------------------------------------
# Merge monthly commodity values into daily dataset
# ------------------------------------------------------------

model_final = model_final.merge(
    commodity_merge,
    on="Month",
    how="left"
)

# Remove temporary month column
model_final = model_final.drop(columns=["Month"])

# ------------------------------------------------------------
# Check result
# ------------------------------------------------------------

print("=" * 70)
print("FINAL DATASET WITH COMMODITY VARIABLES")
print("=" * 70)

print("Shape:", model_final.shape)

print("\nMissing values:")
print(model_final.isna().sum())

print("\nFirst 5 rows:")
display(model_final.head())

print("\nLast 5 rows:")
display(model_final.tail())
```
